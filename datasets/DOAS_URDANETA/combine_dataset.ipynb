```python
import pandas as pd

files_by_year = {
    2014: 'DOAS urdaneta2014_NAAQIS_Data - data_template.csv',
    2015: 'DOAS urdaneta2015_NAAQIS_Data - data_template.csv',
    2016: 'DOAS urdaneta2016_NAAQIS_Data - data_template.csv',
    2017: 'DOAS urdaneta2017_NAAQIS_Data - data_template - Copy - Copy.csv',
    2018: 'DOAS urdaneta2018_NAAQIS_Data - data_template - Copy.csv',
    2019: 'DOAS urdaneta2019_NAAQIS_Data - data_template.csv',
    2020: 'DOAS urdaneta 2020_NAAQIS_Data - data_template.csv',
    2021: 'DOAS urdaneta2021_NAAQIS_Data - data_template - Copy.csv',
}

standard_cols = ['date', 'so2', 'nox', 'o3', 'pm2.5', 'pm10', 'co']
clean_dfs = []

for yr, f in sorted(files_by_year.items()):
    try:
        d = pd.read_csv(f, encoding='utf-8')
    except UnicodeDecodeError:
        d = pd.read_csv(f, encoding='latin1')

    d.columns = [c.strip().lower() for c in d.columns]

    # Fix mislabeled year template in the 2019 file
    if yr == 2019 and d['date'].str.contains('2018').all():
        d['date'] = d['date'].str.replace('2018', '2019')

    d = d[standard_cols].copy()
    d['year'] = yr
    clean_dfs.append(d)

combined_df = pd.concat(clean_dfs, ignore_index=True)
combined_df.to_csv('DOAS_urdaneta_2014_2021_combined.csv', index=False)
```
