```python
import pandas as pd

# Load the merged file
df = pd.read_csv("DOAS_urdaneta_2014_2021_combined.csv")

# Parse date into datetime format
parsed = pd.to_datetime(df["date"], format="%m/%d/%Y")

# Extract month and day
df["month"] = parsed.dt.month
df["day"] = parsed.dt.day

# Reorder columns
ordered_columns = [
    "date",
    "year",
    "month",
    "day",
    "so2",
    "nox",
    "o3",
    "pm2.5",
    "pm10",
    "co",
]
df = df[ordered_columns]

# Save updated dataset
df.to_csv("DOAS_urdaneta_2014_2021_combined.csv", index=False)
```
